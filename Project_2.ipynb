# CREATING MAZE MAP

MAZE MAPPING                                    
"🟩" for state: empty (accessible)    
"🪵" for state: Barrier (not accessible)  
"🪜⬆️" for state: Ladder up (accessible upper floor)    
"🪜⬇️" for state: Ladder down (accessible lower floor)    
"💣" for state: Cannon ball  
"🔫" for state: Cannon   
"📍" for state: Start state  
"🟦" for state: End of the ship (not accessible)  
  
  
"The 🟦 states are there to make the index of each ladder the same on both floors, which the ladder is connecting. With that we make sure we are able to code it."



```python
# ∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞
# Maze Navigation using A* Search
# Description: A* search algorithm to navigate a maze with key states:
#              walls, ladders, cannonballs, and cannons. The goal is to help Luke
#              collect cannonballs, laod cannons while considering the defined rules,
#              and reach the finish position efficiently.
# ∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞

# ====================
# Required Imports
# ====================
import math

# --------------------------
# Maze Representation
# --------------------------

maze_map = [
    ['🟦', '🟦', '🟩', '🪜⬇️', '🟩', '🏴‍☠', '🪵', '🪵', '📍', '🪜⬇️', '🟩', '🟩', '🟦', '🟦'],
    ['🟦', '🟦', '🟦', '🔫🪜⬆️', '🪜⬇️', '💣', '🪜⬇️', '🟩', '🟩', '🪜⬆️', '🟩', '🪜⬇️', '🟦', '🟦'],
    ['🟦', '🟦', '🪜⬇️', '🟩', '🪜⬆️', '🟩', '🪜⬆️', '🪜⬇️', '🟩', '🪜⬇️', '🟩', '🪜⬆️', '🟦', '🟦'],
    ['🟦', '🪵', '🪜⬆️', '🟩', '🪜⬇️', '🪜⬇️', '💣', '🪜⬆️', '🪜⬇️', '🪜⬆️', '🟩', '🔫🪜⬇️', '🟦', '🟦'],
    ['🟦', '🔫🪜⬇️', '🟩','🟩', '🪜⬆️', '🪜⬆️', '🟩', '🟩', '🪜⬆️', '🪜⬇️', '🟩', '🪜⬆️', '🪵', '🟦'],
    ['🪜⬇️', '🪜⬆️', '🪵', '🪜⬇️', '💣', '🪜⬇️', '🟩', '🪜⬇️', '🟩', '🪜⬆️', '🪜⬇️', '🟩', '🔫🪜⬇️', '🟦'],
    ['🪜⬆️', '🟩', '🟩', '🪜⬆️', '🪜⬇️', '🪜⬆️', '🪜⬇️', '🪜⬆️', '🪵', '🪜⬇️', '🪜⬆️', '🪜⬇️', '🪜⬆️', '🪵'],
    ['🟦', '🟩', '🟩', '🪜⬇️', '🪜⬆️', '🟩', '🪜⬆️', '🪜⬇️', '💣', '🪜⬆️', '🪜⬇️', '🪜⬆️', '🟩', '🟦'],
    ['🟦', '🟦', '🟦', '🪜⬆️', '🟩', '🪜⬇️', '🟩', '🪜⬆️', '🟩', '🟩', '🪜⬆️', '🪵', '🟦', '🟦'],
    ['🟦', '🟦', '🟦', '🟦', '🪵', '🪜⬆️', '🟩', '🟩', '🪵', '🟦', '🟦', '🟦', '🟦', '🟦']
]

# Define important states in the maze (📍, 💣, 🔫, 🏴‍☠)
start_state = (0, 8)
cannon_balls = [(3, 6), (7, 8), (5, 4), (1, 5)]
cannons = [(3, 11), (5, 12), (4, 1), (1, 3)]
finish_position = (0, 5)


def is_valid(row, column):
    """
    Check if a given position (row, column) is within the boundaries of the maze.

    Args:
        row (int): Row index of the position.
        column (int): Column index of the position.

    Returns:
        bool: True if the position is within maze boundaries, False otherwise.
    """
    return 0 <= row < len(maze_map) and 0 <= column < len(maze_map[0])

def is_walkable(current_position, previous_position = None, parent_map = False):
    """
    Determine if a move to the current position is allowed in the maze.

    Args:
        current_position (tuple): (row, column) of the current position.
        previous_position (tuple, optional): (row, column) of the previous position. Defaults to None.
        parent_map (bool, optional): If True, relaxes restrictions for reconstructing the final path. Defaults to False.

    Returns:
        bool: True if the move is allowed, False otherwise.
    """
    # Get the current cell from the maze using its row and column
    row, column = current_position
    cell = maze_map[row][column]

    # Define block states
    BLOCKED = {"🟦", "🪵"}

    # Return False if blocked, else True
    if cell in BLOCKED:
      return False

    if parent_map or previous_position is None:
       return True

    # Get the previous cell and calculate vertical movement
    previous_row, previous_column = previous_position
    prev_cell = maze_map[previous_row][previous_column]
    vertical_move = row - previous_row

    # Allowed ladder/cannon moves
    LADDER_UP = {"🪜⬆️", "🔫🪜⬆️"}
    LADDER_DOWN = {"🪜⬇️", "🔫🪜⬇️"}
    ALLOWED_LADDERS = LADDER_UP.union(LADDER_DOWN)

    # Moving up
    if vertical_move == 0:
       return True

    # Moving down
    if vertical_move < 0:
       return prev_cell in LADDER_UP and cell in ALLOWED_LADDERS
    if vertical_move > 0:
       return prev_cell in LADDER_DOWN and cell in ALLOWED_LADDERS

    return False

def heuristic(a, b):
    """
    Calculate the straight-line (Euclidean) distance between two positions in the maze.

    Args:
        a (tuple): (row, column) coordinates of the first position.
        b (tuple): (row, column) coordinates of the second position.

    Returns:
        float: Euclidean distance between positions a and b.
    """
    (x1, y1) = a
    (x2, y2) = b

    return math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)

def reconstruct_path(parent_map, goal_position):
    """
    Reconstruct the path from start to end by following the parent_map.

    Args:
        parent_map (dict): Mapping of each position to its parent position.
        goal_position (tuple): The target position to start reconstructing from.

    Returns:
        list[tuple]: Ordered list of positions from start to end.
    """

    # Initialize empty path
    path = []

    # Start from the goal position
    current_position = goal_position

    # Trace back from end to start using parent map
    while current_position is not None:
         path.append(current_position)
         current_position = parent_map.get(current_position)

    # Reverse to arrange path from start to finish
    path.reverse()

    # Return the reconstructed path
    return path

def state_cost(previous_state, next_state):
    """
    Calculate the movement cost when moving from previous_state to next_state.

    Args:
        previous_state (tuple): (row, column) of the current position.
        next_state (tuple): (row, column) of the next position.
        carrying_ball (bool, optional): Whether Luke is holding a cannonball.

    Returns:
        float: Cost of moving from previous_state to next_state.
    """
    # Extract row and column positions
    previous_row, previous_column = previous_state
    next_row, next_column = next_state

    # Get the maze cell types
    previous_cell  = maze_map[previous_row][previous_column]
    next_cell = maze_map[next_row][next_column]

    BLOCKED = {"🪵", "🟦"} # Impassable cells
    WALKABLE = {"🟩","📍","💣"} # Walkable tiles
    LADDERS = {"🪜⬆️", "🪜⬇️"} # Ladder tiles
    CANNON_LADDERS = {"🔫🪜⬆️", "🔫🪜⬇️"}  # Special ladder with cannon

    # Blocked cell cannot be entered
    if next_cell in BLOCKED:
       return float('inf')  # Infinte cost means inaceesible

    # Accesible tiles
    if next_cell in WALKABLE:
      return 1

    # Ladder or cannon-ladder movement cost
    if next_cell in CANNON_LADDERS:
      return 2 if previous_cell in LADDERS.union(CANNON_LADDERS) else 1

    # Default movement cost
    return 1

# ---------------------------------
# Helper function to explore neighbors
# ---------------------------------
def explore_neighbors(current_position, directions, pending_nodes, to_visit, path_cost, estimated_cost, parent_map, goal, visited):
    """
    Explore valid neighbors from the current node and update costs, parent map,
    and pending nodes for A* pathfinding.
    """

    # Calculate the neighbor's position based on current node and direction
    for vertical_move, horizontal_move in directions:
        neighbor_position = (current_position[0] + vertical_move, current_position[1] + horizontal_move)
        neighbor_row, neighbor_column = neighbor_position

        # Skip neighbors that are out of range
        if not is_valid(neighbor_row, neighbor_column):
          continue

        # Skip invalid or blocked neighbors
        if not is_walkable(neighbor_position, current_position):
          continue

        if neighbor_position in visited:
          continue

        # Compute the cost to reach the neighbor
        cost = path_cost[current_position] + state_cost(current_position, neighbor_position)

        # Update neighbor cost if this path is better than any previous path
        if neighbor_position not in path_cost or cost < path_cost[neighbor_position]:
            path_cost[neighbor_position] = cost
            estimated_cost[neighbor_position] = cost + heuristic(neighbor_position, goal)
            parent_map[neighbor_position] = current_position

            # Add neighbor to pending nodes
            if neighbor_position not in pending_nodes:
                pending_nodes.append(neighbor_position)
                to_visit.add(neighbor_position)

# --------------------------
# Main A* Function
# --------------------------
def astar_function(start_state, cannon_balls, cannons, finish_position):
    """
    Perform A* search to navigate the maze, collect cannonballs, load cannons,
    and finally reach the finish state.

    Args:
        start_state (tuple): (row, column) of the starting position.

    Returns:
        list of lists: All paths taken in order (to each cannonball, cannon, and finish)
    """

    # Initialize path, position
    complete_path = []
    current_position = start_state
    carrying_ball = False

    # Remaining items to collect/visit
    remaining_balls = list(cannon_balls)
    remaining_cannons = list(cannons)

    # Continue until all balls collected, cannons visited, and finish reached
    while remaining_balls or remaining_cannons:

        # Determine next target
        if not carrying_ball and remaining_balls:
            target = remaining_balls[0]
        elif carrying_ball and remaining_cannons:
            target = remaining_cannons[0]
        else:
            target = finish_position

        # Set up data structures needed for A* pathfinding
        visited = set()
        to_visit = {current_position}
        parent_map = {current_position: None}
        path_cost = {current_position: 0}
        estimated_cost = {current_position: heuristic(current_position, target)}

        # Explore nodes until target is reached
        while to_visit:
            # Pick node with lowest f = g + h
            current_position = min(to_visit, key=lambda x: estimated_cost.get(x, float('inf')))
            to_visit.remove(current_position)
            if current_position in visited:
                continue
            visited.add(current_position)

            # Update path and position at target
            if current_position == target:
                path = reconstruct_path(parent_map, current_position)
                complete_path.append(path)

                # Pick up the cannonball
                row, col = current_position
                if maze_map[row][col] == "💣":
                    carrying_ball = True
                    if target in remaining_balls:
                        remaining_balls.remove(target)
                # Load cannon and stop carring ball
                elif maze_map[row][col] in {"🔫🪜⬆️", "🔫🪜⬇️"}:
                    carrying_ball = False
                    if target in remaining_cannons:
                        remaining_cannons.remove(target)

                # Stop the search of previous target
                break

            # Explore neighbors in 4 directions using helper function
            directions = [(-1,0),(1,0),(0,-1),(0,1)]
            explore_neighbors(current_position, directions, pending_nodes=[], to_visit=to_visit,
                              path_cost=path_cost, estimated_cost=estimated_cost,
                              parent_map=parent_map, goal=target, visited=visited)
    return complete_path


# --------------------------
# Final path to finish
# --------------------------
def navigate_finish(start_position, finish_position):
    """
    Find the final path from start_position to finish_position using A*
    """

    # Initialize A* data structures
    visited = set()
    pending_nodes = [start_position]
    parent_map = {start_position: None}
    path_cost = {start_position: 0}
    estimated_cost = {start_position: heuristic(start_position, finish_position)}
    to_visit = set(pending_nodes)

    # Explore nodes until goal is reached
    while pending_nodes:

        # Select the node with lowest total estimated cost
        current_position = min(to_visit, key=lambda x: estimated_cost.get(x, float('inf')))
        to_visit.remove(current_position)

        if current_position in visited:
            continue
        visited.add(current_position)


        # Reconstruct the path when finish is reached
        if current_position == finish_position:
          return reconstruct_path(parent_map, current_position)

        # Explore neighbors in all 4 directions
        directions = [(-1,0),(1,0),(0,-1),(0,1)]
        explore_neighbors(current_position, directions, pending_nodes, to_visit, path_cost,
                          estimated_cost, parent_map, finish_position, visited)

# -------------------------------
# Run program to find final path
# -------------------------------

# Run A* to collect all cannonballs/cannons and find final position
collected_paths = astar_function(start_state, cannon_balls, cannons, finish_position)
last_position = collected_paths[-1][-1]

# Find final path to finish using navigate_finish
final_path_to_finish = navigate_finish(last_position, finish_position)

# Append the final path
collected_paths.append(final_path_to_finish)

# Display all paths
for index, path in enumerate(collected_paths, start = 1):
    print(f"Path {index}: {path}")

```
