**a) (2 marks)**

Load the **Metropolitan Police** crime data for the period **August 2023 - July 2025** into a
single pandas dataframe.

 - The optional outcomes and stop and search data is not needed.

 Write code to remove all variables apart from **Month, Location, LSOA name, and Crime

````python
'''
∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞
 - Question 1(a) Load Metropolitan Police crime data (August 2023 - July 2025)
 - Columns retained: Month, Location, LSOA name, Crime type
∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞
'''
# Import required modules for file handling and data analysis
import zipfile
import os
import pandas as pd

# Extract all files from the ZIP folder to current repository 
with zipfile.ZipFile("Police.zip", "r") as zip_file:
    zip_file.extractall(".")  

# Verify extracted files
print(os.listdir())          

# Initialize a list to store DataFrames
crime_dfs = []

# Collect all folders and subfolders in a list
for subdir, dirs, files in os.walk("."):
    for file in files:
        if file.endswith(".csv") and "metropolitan-street" in file.lower():
            csv_path = os.path.join(subdir, file)
            crime_dfs.append(pd.read_csv(csv_path))  # Read each CSV

# Combine CSVs into one DataFrame
df = pd.concat(crime_dfs, ignore_index=True)

# Keep only the relevant columns for analysis
columns_to_keep = ["Month", "Location", "LSOA name", "Crime type"]
df = df[columns_to_keep]

# Check the loaded data
print(df.head())
print("\nUnique months loaded:", df["Month"].unique())
print("\nTotal rows:", len(df))
````

````python
'''
∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞
Question 3(a): Perceptron Function
∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞∞
'''

# Import Numpy for calculations
import numpy as np

# Define function that computes output from inputs and weights
def perceive(x):
    # Store the weights for each input in a NumPy array
    weights = np.array([0.2, 0.4, 0.3, 0.6, 0.2])
    
    # Compute the combined input value by multiplying the weights
    total_input = np.sum(weights * x)
    
    # Output 1 if perceptron activates, else 0
    return 1 if total_input >= 1 else 0

# Decorative Header
print('=' * 55)
print(f'Perceptron Function'.center(50))
print('=' * 55)

# Test inputs for the perceptron
test_inputs = [
    np.array([0.2, 0.3, 0.1, 0.5, 0.2]),
    np.array([0.5, 0.5, 0.1, 1, 1]),
    np.array([0.9, 0.4, 0.3, 0.8, 0.5])
]

# Display perceptron outputs for each test input
for x in test_inputs:
    print(f"Input: {x}  → Output: {perceive(x)}")

# Decorative Footer
print('=' * 60)
print('Function End'.center(50))
print('=' * 60)
````
